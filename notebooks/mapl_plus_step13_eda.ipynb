```python
import pandas as pd

df1 = pd.read_csv('transaction_1_v5.csv')
df2 = pd.read_csv('transaction_2_v5.csv')
df  = pd.concat([df1, df2], ignore_index=True)
df["Date"] = pd.to_datetime(df["DateTime"]).dt.date
```

```python
df.sort_values("TransactionID").head(50)
```

```python
# Gambaran umum
print("Total transaksi:", len(df))
print("Periode:", df["DateTime"].min(), "→", df["DateTime"].max())
print("\nBranch:", df["Branch"].unique())
print("Jumlah branch:", df["Branch"].nunique())

print("\nJumlah SKU:", df["SKU_ID"].nunique())
print("SKU list:\n", df.groupby(["SKU_ID","SKU","Brand","SKU_Category"]).size().reset_index().drop(columns=0))

print("\nCategory:\n", df["SKU_Category"].value_counts())

print("\nTransaksi per branch:\n", df["Branch"].value_counts())

print("\nPromo vs non-promo:\n", df["IsPromo"].value_counts())
print("% promo:", round(df["IsPromo"].mean()*100, 1), "%")

print("\nDiskon yang ada:", sorted(df["DiscountPercentage"].unique()))
```

```python
import calendar

# Cek hari terakhir yang ada di data Oktober
oct_data = df[df["Date"].dt.month == 10]
print("Hari Oktober yang ada di data:")
print(sorted(oct_data["Date"].dt.day.unique()))

# Bandingkan dengan jumlah hari seharusnya di Oktober 2025
print("\nOktober 2025 seharusnya punya hari:", calendar.monthrange(2025, 10)[1], "hari")
```

```python
df["Date"] = pd.to_datetime(df["Date"])

# Cek jumlah hari unik per bulan
print(df.groupby(df["Date"].dt.month)["Date"].nunique())

# Cek total hari unik tanpa Oktober
df_no_oct = df[df["Date"].dt.month != 10]
print("\nTotal hari tanpa Oktober:", df_no_oct["Date"].nunique())
print("Mei–Sep: 31+30+31+31+30 =", 31+30+31+31+30, "hari")
```

```python
# Cek variasi harga per SKU per branch per hari
check = df.groupby(["Date", "Branch", "SKU_ID"])["DiscountPercentage"].nunique()
print("Maksimal variasi discount dalam 1 hari per SKU per branch:")
print(check.max())  # harusnya = 1
print("\nAda ga yang lebih dari 1 diskon dalam 1 hari?")
print((check > 1).sum())
```
